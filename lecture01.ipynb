- Generators
  - generators can return a sequence of multiple values
  - use the yield keyword instead of return in a function


````python
def squares(n=10):
    for i in range(1, n + 1):
        print(f'[squeres] {i}')
        yield i ** 2

for x in squares():
  print(x)
````


 - Generator expressions
  - Another way to make a generator is by using a generator expression

````python
# equivalent to _make_gen() generator
gen = (x ** 2 for x in range(100))

def _make_gen():
    for x in range(100):
        yield x ** 2

gen = _make_gen()
````

````python
r = sum(x ** 2 for x in range(1,5))
print(r)

# equivalent
r = sum([1, 4, 9, 16])
print(r)
````
